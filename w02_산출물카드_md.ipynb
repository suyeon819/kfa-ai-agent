## 담당 소스 (조 공유용)

- **1. DART 전자공시시스템 – 삼성E&A 재무제표**
  - 수집 방법: 1단계(공식 API), 무료, API 키 필요
  - OpenDART API를 활용하여 삼성E&A(028050)의 재무제표 데이터 수집
  - 기업 고유번호(00126308)를 이용하여 연결(CFS) 기준 재무정보 수집
  - 매출액, 영업이익, 당기순이익, 자산·부채·자본 등 주요 계정과목 수집
  - 사업보고서 및 분기보고서의 보고서 코드를 변경하여 기간별 재무 데이터 수집 가능
  - 조원이 활용할 수 있는 것: 기업 고유번호, 사업연도, 보고서 코드를 변경하면 다른 상장사의 재무제표도 동일한 방식으로 수집 가능
  - 유의사항: 기업별 계정과목 명칭 차이, 분기 누적값과 3개월 값의 구분, 연결·별도 기준 통일 필요
  - 출처: https://opendart.fss.or.kr/

- **2. 한국거래소 정보데이터시스템(KRX Data Marketplace) – 주식시장 데이터**
  - 수집 방법: 3단계(파일 다운로드 및 공식 사이트 조회)
  - 한국거래소에서 제공하는 주식시장 데이터를 활용하여 상장기업의 시장정보 수집
  - 종목별 주가, 시가총액, 거래량, PER, PBR, 배당수익률 등 조회
  - 주가 추이 분석, 동종기업 비교 및 밸류에이션 산출에 활용
  - 조원이 활용할 수 있는 것: 종목코드와 조회 기간을 변경하여 다양한 상장기업의 주가 및 주요 시장지표 비교 가능
  - 유의사항: 주가 기준일을 통일하고, PER·PBR의 산출 기준 및 재무정보 기준일 확인 필요
  - 출처: https://data.krx.co.kr/
